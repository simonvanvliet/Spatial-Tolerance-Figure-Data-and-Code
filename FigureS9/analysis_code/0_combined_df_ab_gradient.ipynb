# Aggregate Fluorescence Intensity Data for AB gradient experiment

## Overview
This notebook aggregates fluorescence intensity data from multiple experimental replicates and positions for the wt strain, where AB were supplied only from PA side

## Workflow
1. **Input**: Data comes from  `ToleranceAssay` study component in BioImageArchive repo
2. **Data Collection**: Iterates over all replicates and positions and loads `single_cell_props.csv` files.
3. **Output**: Saves the combined dataframe as `0_combined_df_ab_gradient.csv`


```python
# set path to BioImageArchive data directory:
data_archive_path = '/Volumes/ScientificData/Users/Giulia(botgiu00)/Papers/bottacin2026/BioImageArchive/'

# input and output paths relative to BioImageArchive location
input_dir = os.path.join(data_archive_path, 'ToleranceAssay/S9_wt_AB-gradient')
output_dir = '0_combined_df_ab_gradient.csv'

# Create empty list to store dataframes
all_data = []

# Loop through replicate folders within the rhlA folder
for replicate_folder in os.listdir(input_dir):
    replicate_path = os.path.join(input_dir, replicate_folder)
    
    # Ensure we are processing replicate directories
    if os.path.isdir(replicate_path) and replicate_folder.startswith('replicate'):
        
        # Loop through position folders within each replicate folder
        for pos_folder in os.listdir(replicate_path):
            if pos_folder.startswith('pos'):
                pos_path = os.path.join(replicate_path, pos_folder, 'single_cell_props.csv')
                
                # Check if CSV exists and load data
                if os.path.isfile(pos_path):
                    df = pd.read_csv(pos_path)
                    
                    # Add 'rhlA', 'replicate', and 'pos' columns
                    df.insert(1, 'strain', 'wt')
                    df.insert(2, 'replicate', replicate_folder)
                    df.insert(3, 'pos', pos_folder)
                    
                    # Append to the list
                    all_data.append(df)

# Merge into a single dataframe and save at output_dir
if all_data:
    combined_df_ab_gradient = pd.concat(all_data, ignore_index=True)
    combined_df_ab_gradient.to_csv(output_dir, index=False)

combined_df_ab_gradient.head()
```
