# Plot Figure 4A: Predicted survival mixed chambers

Plots predicted survival in mixed chambers 

This code is provided here for completeness only, full code containing model fitting and parameter inference is provided in complimentary [Github Repository](https://github.com/simonvanvliet/SpatialToleranceModel) 

```python
import numpy as np
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
output_df = pd.read_csv("cluster_size_vs_survival.csv")

# -----------------------------
# Plot style
# -----------------------------
font_size = 8
plt.rcParams.update({
    'font.size': font_size,
    'font.family': 'Arial',
    'axes.labelsize': font_size,
    'axes.titlesize': font_size,
    'xtick.labelsize': font_size,
    'ytick.labelsize': font_size,
    'legend.fontsize': font_size,
    'pdf.fonttype': 42,
    'ps.fonttype': 42
})


# -----------------------------
# Plot line 
# -----------------------------


fig, ax = plt.subplots(figsize=(6/2.5, 6/2.5))
sns.lineplot(data=output_df, x="cluster size", y="survival probability", marker="o", ax=ax, color="black", markersize=4, legend=0, markeredgewidth=0)
ax.set_xlabel("Patch size (μm)")
ax.set_ylabel("Survival probability")
ax.set_xlim(0, 30)
ax.set_ylim(-0.0005, 0.004)  # Adjust as needed
ax.set_yticks(np.linspace(0, 0.004, 5))


fig, ax = plt.subplots(figsize=(6/2.5, 6/2.5))
sns.lineplot(data=output_df, x="cluster size", y="MSSL", marker="o", ax=ax, color="black", markersize=4, legend=0, markeredgewidth=0)
ax.set_xlabel("Patch size (μm)")
ax.set_ylabel("MSSL")
ax.set_xlim(0, 30)
ax.set_ylim(0, 1)  # Adjust as needed


fig, ax = plt.subplots(figsize=(6/2.5, 6/2.5))
sns.lineplot(data=output_df, x="MSSL", y="survival probability", marker="o", ax=ax, color="black", markersize=4, legend=0, markeredgewidth=0)
ax.set_xlabel("Multiscale spatial segregation level (MSSL)")
ax.set_ylabel("Survival probability")
ax.set_xlim(0, 0.8)
ax.set_ylim(0, 0.0012)  # Adjust as needed
ax.set_yticks(np.linspace(0, 0.0012, 5))


# -----------------------------
# Calculate Spearman correlation
# -----------------------------
from scipy.stats import spearmanr
rho, pval = spearmanr(output_df['MSSL'], output_df['survival probability'])
r2 = rho**2

print(f"Spearman correlation:")
print(f"  ρ = {rho:.3f}")
print(f"  R² = {r2:.3f}")
print(f"  p-value = {pval:.4g}")

```
