```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

This script aggregates fluorescence intensity data from multiple replicates and positions for the `wt` strain. It performs the following steps:
- Iterates through all `replicate*` folders inside the input directory.
- Within each replicate, it searches for `pos*` folders and reads the `phase/fluo_intensities.csv` file if present.
- Adds metadata columns: `strain` (set to `'∆rhlA'`), `replicate`, and `pos` to each loaded dataframe.
- Appends all dataframes into a list and concatenates them into a single dataframe.
- Saves the combined dataframe to the specified output path as `0_combined_df_wt.csv`.



```python
# Define input and output files for wt
input_dir = '/Volumes/ScientificData/Users/Giulia(botgiu00)/Papers/bottacin2025/Image_Data/Figure2/2S_pa_flush'
output_dir = '/Volumes/ScientificData/Users/Giulia(botgiu00)/Papers/bottacin2025/Figure_Data/Figure2/2S_pa_flush/0_combined_df_pa_flush.csv'

# Create empty list to store dataframes
all_data = []

# Loop through replicate folders within the rhlA folder
for replicate_folder in os.listdir(input_dir):
    replicate_path = os.path.join(input_dir, replicate_folder)
    
    # Ensure we are processing replicate directories
    if os.path.isdir(replicate_path) and replicate_folder.startswith('replicate'):
        
        # Loop through position folders within each replicate folder
        for pos_folder in os.listdir(replicate_path):
            if pos_folder.startswith('pos'):
                pos_path = os.path.join(replicate_path, pos_folder, 'phase', 'fluo_intensities.csv')
                
                # Check if CSV exists and load data
                if os.path.isfile(pos_path):
                    df = pd.read_csv(pos_path)
                    
                    # Add 'rhlA', 'replicate', and 'pos' columns
                    df.insert(1, 'strain', 'wt')
                    df.insert(2, 'replicate', replicate_folder)
                    df.insert(3, 'pos', pos_folder)
                    
                    # Append to the list
                    all_data.append(df)

# Merge into a single dataframe and save at output_dir
if all_data:
    combined_df_pa_flush = pd.concat(all_data, ignore_index=True)
    combined_df_pa_flush.to_csv(output_dir, index=False)

combined_df_pa_flush.head()
```
