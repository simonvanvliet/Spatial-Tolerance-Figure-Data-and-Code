```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

### prepare background dataframe

```python
# Define input and output files for background
input_dir = '/Volumes/ScientificData/Users/Giulia(botgiu00)/Paper/Image_Data/Figure2/background'
output_dir = '/Volumes/ScientificData/Users/Giulia(botgiu00)/Paper/Image_Data/Figure2/background/0_combined_df_bg.csv'

# Create empty list to store dataframes
all_data = []

# Loop through replicate folders within the wt folder
for replicate_folder in os.listdir(input_dir):
    replicate_path = os.path.join(input_dir, replicate_folder)
    
    # Ensure we are processing replicate directories
    if os.path.isdir(replicate_path) and replicate_folder.startswith('replicate'):
        
        # Loop through position folders within each replicate folder
        for pos_folder in os.listdir(replicate_path):
            if pos_folder.startswith('pos'):
                pos_path = os.path.join(replicate_path, pos_folder, 'phase', 'fluo_intensities.csv')
                
                # Check if CSV exists and load data
                if os.path.isfile(pos_path):
                    df = pd.read_csv(pos_path)
                    
                    # Add 'strain', 'replicate', and 'pos' columns
                    df.insert(1, 'strain', 'wt')
                    df.insert(2, 'replicate', replicate_folder)
                    df.insert(3, 'pos', pos_folder)
                    
                    # Append to the list
                    all_data.append(df)

# Merge into a single dataframe, filter, and save
if all_data:
    combined_df_bg = pd.concat(all_data, ignore_index=True)

    # Filter for frame_number between 0 and 10 (inclusive)
    combined_df_bg = combined_df_bg[combined_df_bg['frame_number'].between(0, 10)]

    # Save filtered dataframe
    combined_df_bg.to_csv(output_dir, index=False)
```
